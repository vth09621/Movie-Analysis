```python
import pandas as pd
import numpy as np
movie = pd.read_csv("../data_sets/project_3(Movies)/movies.csv")
rating = pd.read_csv("../data_sets/project_3(Movies)/ratings.csv")
user = pd.read_csv("../data_sets/project_3(Movies)/users.csv")
rating["rating_date"] = pd.to_datetime(rating["rating_date"])
user["signup_date"] = pd.to_datetime(user["signup_date"])
```

```python
df = [("movie",movie),("rating",rating),("user", user)]
for name,df in df:
    print(f"============{name}========")
    print(df.head())
    print(df.tail())
    print(df.describe())
    print(df.info())


df = [("movie",movie),("rating",rating),("user", user)]
for name,df in df:
    print(f"==========={name}============")
    print(df.isnull().sum())
    print(df.duplicated().sum())    
    
```

```python
# Merge
merged = movie.merge(
    rating,
    on="movie_id"
)

merged = merged.merge(
    user,
    on="user_id"
)
# export
merged.to_csv(
    "movie_cleaned.csv",
    index=False
)

print(merged.head())



print("Duplicate columns = ", merged.columns.duplicated().sum())
print(merged.columns)
print("Dulpicate = ", merged.duplicated().sum())
print(merged.dtypes)
```
